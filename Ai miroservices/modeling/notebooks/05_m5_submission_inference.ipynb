# 05 M5 Submission Generation

This notebook generates Kaggle submission CSV files for the M5 competition using the current classical forecasting family in this repo.

What Kaggle expects:
- one `id` column
- forecast columns `F1` to `F28`
- both `_validation` and `_evaluation` rows
- same row order and shape as `sample_submission.csv`

Important limitation:
- `ARIMA`, `SARIMA`, and `ETS` still have to be fit per time series before forecasting. There is no separate offline training artifact for them.
- `XGBOOST` and `CATBOOST` are not included here because they need training on M5-compatible features or a compatible pretrained model artifact. Without that, they cannot generate valid M5 submission forecasts.

```python
from pathlib import Path
import pandas as pd

M5_DIR = Path('/Users/k.e.oshada/Documents/OptiWMS/Ai miroservices/external-data/m5-forecasting-accuracy')
OUT_DIR = Path('/Users/k.e.oshada/Documents/OptiWMS/Ai miroservices/modeling/outputs/reports')
SCRIPT = Path('/Users/k.e.oshada/Documents/OptiWMS/Ai miroservices/modeling/scripts/m5_submission_inference.py')

sample_submission = pd.read_csv(M5_DIR / 'sample_submission.csv')
print('sample_submission shape:', sample_submission.shape)
print('columns:', sample_submission.columns.tolist()[:6], '...', sample_submission.columns.tolist()[-3:])
sample_submission.head()
```
