# 05C Transfer Diagnostics

This notebook opens transfer prediction outputs and builds diagnostics by horizon, category, and family.

```python
from pathlib import Path
import pandas as pd
import subprocess
import sys

ROOT = Path('/Users/k.e.oshada/Documents/OptiWMS/Ai miroservices/modeling')
REPORTS_DIR = ROOT / 'outputs' / 'reports'
SCRIPT = ROOT / 'scripts' / 'transfer_diagnostics.py'

PREDICTION_TAG = 'portable_m5_transfer_auto_capped'
PREDICTIONS = REPORTS_DIR / f'{PREDICTION_TAG}_predictions.csv'
DIAG_TAG = f'{PREDICTION_TAG}_diag'
```

## Build Diagnostics

```python
cmd = [
    sys.executable,
    str(SCRIPT),
    '--predictions', str(PREDICTIONS),
    '--tag', DIAG_TAG,
]
print('RUNNING', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

## By Horizon

```python
pd.read_csv(REPORTS_DIR / f'{DIAG_TAG}_by_horizon.csv')
```

## By Category

```python
pd.read_csv(REPORTS_DIR / f'{DIAG_TAG}_by_category.csv')
```
